```python
# Amazon Dataset — 10 Basic Matplotlib Charts
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Load dataset
df = pd.read_csv("Amazon_clear_Data.csv")

# Display first 5 rows
df.head()
```

```python
# 1. Price Distribution — Histogram

plt.figure(figsize=(8, 5))
plt.hist(df["Price"].dropna(), bins=20, edgecolor="black")
plt.title("Distribution of Product Prices")
plt.xlabel("Price")
plt.ylabel("Number of Products")
plt.grid(axis="y", alpha=0.3)
plt.show()
```

```python
# 2. Rating Distribution — Histogram

plt.figure(figsize=(8, 5))
plt.hist(df["Rating"].dropna(), bins=10, edgecolor="black")
plt.title("Distribution of Product Ratings")
plt.xlabel("Rating")
plt.ylabel("Number of Products")
plt.grid(axis="y", alpha=0.3)
plt.show()
```

```python
# 3. Number of Products by Brand — Bar Chart

brand_counts = df["Brand"].value_counts()

plt.figure(figsize=(10, 5))
plt.bar(brand_counts.index.astype(str), brand_counts.values)
plt.title("Number of Products by Brand")
plt.xlabel("Brand")
plt.ylabel("Number of Products")
plt.xticks(rotation=45)
plt.grid(axis="y", alpha=0.3)
plt.show()
```

```python
# 4. Average Price by Brand — Bar Chart

avg_price_brand = df.groupby("Brand")["Price"].mean().sort_values(ascending=False)

plt.figure(figsize=(10, 5))
plt.bar(avg_price_brand.index.astype(str), avg_price_brand.values)
plt.title("Average Price by Brand")
plt.xlabel("Brand")
plt.ylabel("Average Price")
plt.xticks(rotation=45)
plt.grid(axis="y", alpha=0.3)
plt.show()
```

```python
# 5. Average Rating by Brand — Bar Chart

avg_rating_brand = df.groupby("Brand")["Rating"].mean().sort_values(ascending=False)

plt.figure(figsize=(10, 5))
plt.bar(avg_rating_brand.index.astype(str), avg_rating_brand.values)
plt.title("Average Rating by Brand")
plt.xlabel("Brand")
plt.ylabel("Average Rating")
plt.xticks(rotation=45)
plt.grid(axis="y", alpha=0.3)
plt.show()
```

```python
# 6. Number of Products by RAM — Bar Chart

ram_counts = df["Ram(GB)"].value_counts().sort_index()

plt.figure(figsize=(8, 5))
plt.bar(ram_counts.index.astype(str), ram_counts.values)
plt.title("Number of Products by RAM")
plt.xlabel("RAM (GB)")
plt.ylabel("Number of Products")
plt.grid(axis="y", alpha=0.3)
plt.show()
```

```python
# 7. Number of Products by SSD Storage — Bar Chart

ssd_counts = df["Ssd_Storage(GB)"].value_counts().sort_index()

plt.figure(figsize=(8, 5))
plt.bar(ssd_counts.index.astype(str), ssd_counts.values)
plt.title("Number of Products by SSD Storage")
plt.xlabel("SSD Storage (GB)")
plt.ylabel("Number of Products")
plt.grid(axis="y", alpha=0.3)
plt.show()
```

```python
# 8. Number of Products by Processor — Bar Chart

processor_counts = df["Processor"].value_counts()

plt.figure(figsize=(8, 5))
plt.bar(processor_counts.index.astype(str), processor_counts.values)
plt.title("Number of Products by Processor")
plt.xlabel("Processor")
plt.ylabel("Number of Products")
plt.xticks(rotation=45)
plt.grid(axis="y", alpha=0.3)
plt.show()
```

```python
# 9. Number of Products by Color — Bar Chart

color_counts = df["Color"].value_counts()

plt.figure(figsize=(10, 5))
plt.bar(color_counts.index.astype(str), color_counts.values)
plt.title("Number of Products by Color")
plt.xlabel("Color")
plt.ylabel("Number of Products")
plt.xticks(rotation=45)
plt.grid(axis="y", alpha=0.3)
plt.show()
```

```python
# 10. Price vs Rating — Scatter Plot

plt.figure(figsize=(8, 5))
plt.scatter(df["Price"], df["Rating"], alpha=0.6)
plt.title("Price vs Rating")
plt.xlabel("Price")
plt.ylabel("Rating")
plt.grid(alpha=0.3)
plt.show()
```
