```python
## 1. Class Name: Person → Student
Problem
Create a parent class named Person and a child class named Student.
Explanation
The Person class should store:
• name
• age
The Student class should inherit from Person and add:
• marks
Create a method in Student to display the student's name, age, and marks.
```

```python
class Person:
    def __init__(self,name,age):
        self.Name=name
        self.Age=age
        
    def display_info(self):
        print("student name  is",self.Name)
        print("student age is ",self.Age)


class Student(Person):
    def __init__(self,name,age,marks):
        super().__init__(name,age)
        self.Marks=marks

    def display_all_info(self):
        
        self.display_info()
        print("student marks is",self.Marks)
        #display_info()



s=Student("gokul",21,89)
s.display_all_info()
```

```python
## 4. Class Name: Employee → Manager
Problem
Create a parent class named Employee.
Explanation
The Employee class should store:
• emp_id
• salary
Create child class Manager.
Add:
• department
Display all employee information.
```

```python
class Employee:
    def __init__(self,emp_id,salary):
        self.Employee_id=emp_id
        self.Salary=salary


    def display_emp_info(self):
        print("employee id:",self.Employee_id)
        print("employee salary",self.Salary)


class Manager(Employee):
    def __init__(self,emp_id,salary,department):
        super().__init__(emp_id,salary)
        self.Department=department


    def display_all_info(self):
        
        self.display_emp_info()
        print("employee department",self.Department)


emp=Manager("@gokul12",100000,"software engineeer")
emp.display_all_info()
    
```

```python
## 5. Class Name: Shape → Circle
Problem
Create a parent class named Shape.
Explanation
Create a method:
• area()
Create child class:
• Circle
Store:
• radius
Override area() to calculate the area of a circle.
```

```python
import math
class Shape:
    def area(self):
        pass

class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius
    def area(self):
        return math.pi * self.radius * self.radius


circle = Circle(5)
print("Area of Circle:", circle.area())
    
```

```python
## 6. Class Name: BankAccount → SavingsAccount
Problem
Create a parent class named BankAccount.
Explanation
Provide methods:
• deposit()
• withdraw()
Create subclass:
• SavingsAccount
Add:
• calculate_interest()
```

```python
class BankAccount:
    def __init__(self,ac_no,balance):
        self.Account_number=ac_no
        self.Balance=balance

    def deposit(self,new_amt):
        if new_amt >0:
            self.Balance += new_amt
            print("your amount deposit successfully:",self.Balance)

        else:
            print("please enter amount more then 0")


    def withdraw(self,amt):

        if amt <= self.Balance:
            self.Balance-=amt
            print("money withdraw successfully",amt)
            print("remaining balance :",self.Balance)

        else:
            print("please enter valid amount")

    def display_info(self):
        print("your account number is:",self.Account_number)
        print("your balance is :",self.Balance)


class SavingsAccount(BankAccount):
    def __init__(self,ac_no,balance,interst_rate):
        super().__init__(ac_no,balance)
        self.interst_rate=interst_rate


    def calculate_interest(self):
        interst=self.Balance * self.interst_rate /100
        print("your interst is ",interst)
        


    #def display_all_info(self)


bank=BankAccount(12345,5000)
bank.deposit(500)
bank.withdraw(200)

bank=SavingsAccount(54624,4000000,5)
bank.calculate_interest()
bank.display_info()
        
        
    
    
        
    
    
        
```

```python
## 7. Class Name: Product → Electronics
Problem
Create a parent class named Product.
Explanation
Store:
• price
Create method:
• discount()
Create child class:
• Electronics
Add:
• warranty
Display all information.
```

```python
class Product:
    def __init__(self,price):
        self.Price=price


    def discount(self,percent):
        discount=self.Price * percent /100
        final_discount=self.Price - discount
        print("discount is ",discount)
        print("final_discount is ",final_discount)



class Electronics(Product):
    def __init__(self,price, warranty):
        super().__init__(price)
        self.warranty=warranty


    def display_all_info(self):
        print("price is",self.Price)
        self.discount(10)
        print("warranty",self.warranty)


#p=Product(500)
#p.discount()
e=Electronics(500,3)
e.display_all_info()

        
    
        
        
```

```python
## 12. Class Name: Account → LoanAccount
Problem
Create a parent class named Account.
Explanation
Store:
• balance
Create subclass:
• LoanAccount
Add:
• calculate_emi()
```

```python
class Account:
    def __init__(self,balance):
        self.Balance=balance

class Loneaccount(Account):
    def __init__(self,balance,lone_amount,month):
        super().__init__(balance)
        self.Lone_amount=lone_amount
        self.Month=month
    def calculate_emi(self,emi):
        emi=self.Lone_amount / self.Month
        print("your emi is :",emi)

    def display_info(self):
        print("your balance:",self.Balance)
        print("your lone amount is :",self.Lone_amount)
        print("pending month is :",self.Month)

account=Loneaccount(100,12000,6)
account.calculate_emi(12000)
account.display_info()
    
```

```python
## 13. Class Name: Employee → Developer
Problem
Create a parent class named Employee.
Explanation
Create method:
• get_salary()
Create subclass:
• Developer
Add:
• bonus
Display total salary including bonus.
```

```python
class Employee:
    def __init__(self,name,salary):
        self.Name=name
        self.Salary=salary
    def get_salary(self):
        print("your salary:",self.Salary)

class Developer(Employee):
    def __init__(self,name,salary,bonus):
        super().__init__(name,salary)
        self.Bonus=bonus

    def display_total_salary(self):
        total_salary=self.Salary + self.Bonus
        print("your name :",self.Name)
        print("your salary :",self.Salary)
        print("total salary including bonus:",total_salary)

employee=Developer("gokul",100000,10000)
employee.display_total_salary()
        
    
```
