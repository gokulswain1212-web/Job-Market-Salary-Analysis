```python
##1. Class Name: Shape
Problem
Create an abstract class named Shape.
Explanation
A shape can have different formulas for area and perimeter. Therefore, create two abstract
methods:
• area()
• perimeter()
Create subclasses like Circle and Rectangle and implement both methods.
```

```python
from abc import ABC, abstractmethod
import math

class Shape(ABC):

    @abstractmethod
    def area(self):
        pass

    @abstractmethod
    def perimeter(self):
        pass



class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return math.pi * self.radius ** 2

    def perimeter(self):
        return 2 * math.pi * self.radius


class Rectangle(Shape):
    def __init__(self, length, width):
        self.length = length
        self.width = width

    def area(self):
        return self.length * self.width

    def perimeter(self):
        return 2 * (self.length + self.width)



circle = Circle(5)
rectangle = Rectangle(4, 6)

print("Circle Area =", circle.area())
print("Circle Perimeter =", circle.perimeter())

print("Rectangle Area =", rectangle.area())
print("Rectangle Perimeter =", rectangle.perimeter())
    
```

```python
## 3. Class Name: Animal
Problem
Create an abstract class named Animal.
Explanation
Every animal makes a sound, but the sound depends on the animal
Create an abstract method:
• make_sound()
Create subclasses:
• Dog
• Cat
Each should implement its own sound.
```

```python
class Animal(ABC):

    @abstractmethod
    def make_sound(self):
        pass   


class Dog(Animal):

    def make_sound(self):
        print("bhouuuuuuu !bhouuuuuu")


class Cat(Animal):

    def make_sound(self):
        print("meow !meow")


animal=Dog()
animal.make_sound()
animal=Cat()
animal.make_sound()
```

```python
## 4. Class Name: Payment
Problem
Create an abstract class named Payment.
Explanation
Every payment processes money differently.
Create an abstract method:
• process_payment(amount)
Create subclasses:
• CreditCard
• Cash
Each should process the payment in its own way.
```

```python
class Payment(ABC):


    @abstractmethod
    def process_payment(self,amount):
        pass


class CreditCard(Payment):


    def process_payment(self,amount):
        print("proccessing your CreditCard card amount",amount)

class Cash(Payment):

    def process_payment(self,amount):
        print("proccessing your cash amount",amount)



payment=CreditCard()
payment.process_payment(100)
payment=Cash()
payment.process_payment(500)
    
```

```python
## 6. Class Name: BankAccount
Problem
Create an abstract class named BankAccount.
Explanation
Different bank accounts calculate interest differently.
Create an abstract method:
• calculate_interest()
Create:
• SavingsAccount
Implement the interest calculation.
```

```python
class BankAccount(ABC):

    @abstractmethod
    def calculate_interes(self):
        pass

class SavingsAccount(BankAccount):

    def __init__(self,balance,rate):
        self.Balance=balance
        self.Rate=rate

    def calculate_interes(self):
        interest=(self.Balance * self.Rate)/100
        print("your interest is ",interest)

    def display_info(self):
        print("your balance is",self.Balance)
        print("your rate is ",self.Rate)


bank=SavingsAccount(10000,4)
bank.calculate_interes()
bank.display_info()
```

```python
## 7. Class Name: Device
Problem
Create an abstract class named Device.
Explanation
Every electronic device can be turned on and off.
Create abstract methods:
• turn_on()
• turn_off()
Implement them in a subclass such as Laptop or TV.
```

```python
class Device(ABC):

    @abstractmethod
    def turn_on(self):
        pass

    @abstractmethod
    def turn_off(self):
        pass


class Laptop(Device):

    def turn_on(self):
        print("when you press on button then your laptop turn on")

    def turn_off(self):
        print("when you press off button then your laptop turn off")


class TV(Device):

    def turn_on(self):
        print("when you press on button then your tv turn on")

    def turn_off(self):
        print("when you press off button then your tv turn off")


device=Laptop()
device.turn_on()
device.turn_off()
device=TV()
device.turn_on()
device.turn_off()


    
```

```python
## 8. Class Name: Employee
Problem
Create an abstract class named Employee.
Explanation
Employees receive bonuses based on their roles.
Create an abstract method:-
• calculate_bonus()
Create subclass:
• Manager
Implement the bonus calculation.
```

```python

class Employee(ABC):

    @abstractmethod
    def calculate_bonus(self):
        pass



class Manager(Employee):

    def __init__(self, salary):
        self.salary = salary

    def calculate_bonus(self):
        bonus = self.salary * 20 
        print("Manager Bonus =", bonus)



manager = Manager(50000)
manager.calculate_bonus()
```

```python
## 9. Class Name: Transport
Problem
Create an abstract class named Transport.
Explanation
Every transport moves differently and uses different fuel.
Create abstract methods:
• move()
• fuel_type()
Create subclasses like:
• Car
• Train
```

```python
from abc import ABC, abstractmethod
class Transport(ABC):

    @abstractmethod
    def move(self):
        pass

    @abstractmethod
    def fuel_type(self):
        pass


class Car(Transport):

    def move(self):
        print("car move on road")

    def fuel_type(self):
        print("car use petrol as fuel")


class Train(Transport):

    def move(self):
        print("train move on train line")

    def fuel_type(self):
        print("train use electricity and diesel as fule")


tr=Car()
tr.move()
tr.fuel_type()
tr=Train()
tr.move()
tr.fuel_type()
```
