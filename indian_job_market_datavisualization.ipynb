```python
import pandas as pd
df=pd.read_csv(r"C:\Users\LENOVO\Downloads\indian_job_market_clean_data.csv")
df.head(2)
```

```python
df.columns
```

```python
# Which locations have the most jobs
import matplotlib.pyplot as plt

top_titles = df["Title"].value_counts().head(10)

plt.figure(figsize=(10, 6))
plt.bar(top_titles.index, top_titles.values)

plt.xlabel("Job Title")
plt.ylabel("Number of Jobs")
plt.title("Top 10 Job Titles")

plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```

```python
top_locations = df["Location"].value_counts().head(10)

plt.figure(figsize=(10, 6))
plt.barh(top_locations.index, top_locations.values)

plt.xlabel("Number of Jobs")
plt.ylabel("Location")
plt.title("Top 10 Job Locations")

plt.gca().invert_yaxis()
plt.tight_layout()
plt.show()
```

```python
#Average Minimum vs Maximum Salary
salary = df[["Maximum_salary", "Minimum_salary"]].mean()

plt.figure(figsize=(8, 5))
plt.bar(salary.index, salary.values)

plt.xlabel("Salary Type")
plt.ylabel("Average Salary")
plt.title("Average Minimum vs Maximum Salary")

plt.tight_layout()
plt.show()
```

```python
#Jobs by Minimum Experience
experience = df["Minimum_experience"].value_counts().sort_index()

plt.figure(figsize=(10, 6))
plt.bar(experience.index.astype(str), experience.values)

plt.xlabel("Minimum Experience (Years)")
plt.ylabel("Number of Jobs")
plt.title("Jobs by Minimum Experience")

plt.tight_layout()
plt.show()
```

```python
# Aggregate Rating Distribution
plt.figure(figsize=(8, 5))

plt.hist(
    df["Aggregaterating"].dropna(),
    bins=10
)

plt.xlabel("Aggregate Rating")
plt.ylabel("Number of Jobs")
plt.title("Distribution of Aggregate Ratings")

plt.tight_layout()
plt.show()
```
