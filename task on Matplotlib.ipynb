# 1. Create a simple line plot using:
# o X = [1, 2, 3, 4, 5]
# o Y = [10, 20, 15, 25, 30]

```python
import matplotlib.pyplot as plt
import numpy as np
X = [1, 2, 3, 4, 5]
Y = [10, 20, 15, 25, 30]

plt.plot(X,Y)
plt.show()
```

## 2. Create a line plot and add an appropriate title.

```python
X = [1, 2, 3, 4, 5]
Y = [10, 20, 15, 25, 30]

plt.plot(X,Y)
plt.title("simple plot")
plt.show()
```

## 3. Create a line plot and label both the X-axis and Y-axis.

```python
X = [1, 2, 3, 4, 5]
Y = [10, 20, 15, 25, 30]

plt.plot(X,Y)
plt.title("simple plot")
plt.xlabel("x label")
plt.ylabel("y label")
plt.show()
```

## 4. Plot two lines on the same graph and add a legend to identify them.

```python
X = [1, 2, 3, 4, 5]
Y = [10, 20, 15, 25, 30]

plt.plot(X,Y)
plt.title("simple plot")
plt.xlabel("x label")
plt.ylabel("y label")

plt.plot(X,  label="Line 1")
plt.plot(X,  label="Line 2")
plt.show()
```

# 5. Change the line color, line style, and marker style of a line plot.

```python
plt.plot(X,Y,
         color="red",
         linewidth='1.2',
         marker="*"
        )
plt.show()
```

# 7. Create a bar chart showing the sales of 5 different products.

```python
products = ["laptop", "mobile", "charger", "computer", "cpu"]
sales = [120, 150, 90, 180, 130]

plt.bar(products, sales,color="green")

plt.xlabel("Products")
plt.ylabel("Sales")
plt.title("Sales of 5 Different Products")

plt.show()
```

## 8. Create a horizontal bar chart for the same product-sales data.

```python
products = ["laptop", "mobile", "charger", "computer", "cpu"]
sales = [120, 150, 90, 180, 130]

plt.barh(products, sales,color="green")

plt.xlabel("Products")
plt.ylabel("Sales")
plt.title("Sales of 5 Different Products")

plt.show()
```

##  12.Create a plot containing two subplots:
# o First subplot → line plot
# o Second subplot → scatter plot

```python


fig, axes = plt.subplots(2, dpi=100, figsize=(5, 5))

fig.set_facecolor("pink")

fig.suptitle(
    "Figure Title",
    fontsize=15,
    fontweight="bold"
)

categories = ["A", "B", "C", "D", "E"]
values = [10, 25, 15, 30, 20]

axes[1].plot(
    categories,
    values,
    color="orange"
)

axes[1].set_title("Bar Plot")
axes[1].set_xlabel("Categories", fontsize=12)
axes[1].set_ylabel("Values", fontsize=12)

axes[1].grid(
    True,
    axis="y",
    linestyle="--",
    alpha=0.9
)

Age = [21, 54, 32, 52, 30, 35, 40, 28, 45, 38]
salary = [120000, 1541225, 45100, 45000, 50000,
          80000, 90000, 60000, 110000, 75000]

axes[0].scatter(Age, salary, color="black")
axes[0].set_title("Scatter Plot")
axes[0].set_xlabel("Age")
axes[0].set_ylabel("Salary")
axes[0].grid(True, linestyle="--", alpha=0.5)

plt.tight_layout()

plt.show()
```
