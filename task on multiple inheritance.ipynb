## 11. Create two parent classes with static methods. Create a child class that inherits from
## both and calls the static methods.

```python

class Parent1:
    @staticmethod
    def show_message1():
        print("Static method from Parent1")
class Parent2:
    @staticmethod
    def show_message2():
        print("Static method from Parent2")

class Child(Parent1, Parent2):
    @staticmethod
    def call_methods():
        Parent1.show_message1()
        Parent2.show_message2()

Child.call_methods()
```

## 19. Create two abstract classes Shape and Color. Create a concrete class that inherits fromboth and implements abstract methods.

```python
from abc import ABC, abstractmethod
class Shape(ABC):
    @abstractmethod
    def draw(self):
        pass

class Color(ABC):
    @abstractmethod
    def fill(self):
        pass

class Circle(Shape, Color):
    def draw(self):
        print("Drawing a Circle")

    def fill(self):
        print("Filling the Circle with Red color")
c = Circle()
c.draw()
c.fill()
```

## 20. Define abstract class Payment and abstract class Notification. Create classOnlinePayment that inherits from both.

```python
class Payment(ABC):
    def __init__(self,password,balance):
        self.__Password=password
        self.__Balance=balance
        
    def get_password(self):
        return self.__Password

    def set_password(self,password):
        password=int(input("enter your password:"))
        if password == self.__Password:
            print("correct password",password)

        else:
            print("please correct password")

    def get_balance(self):
        return self.__Balance

    def set_balance(self,amt):
        if amt >= 0:
            self.__Balance = amt
            print("balance update",self.__Balance)


    @abstractmethod
    def make_payment(self):
        pass

    

class Notification(ABC):
    @abstractmethod
    def send_notification(self, message):
        pass



class OnlinePayment(Payment,Notification):
    def __init__(self,password,balance,):
        super().__init__(password,balance)
        #self.Amount=amt


    def make_payment(self,amt):
        if amt <= self.get_balance():
            new_balance = self.get_balance() - amt
            self.set_balance(new_balance)
            print("Payment of", amt, "successful.")
        else:
            print("Insufficient balance.")
    


    def send_notification(self, message):
        print("Notification:",message)



    def display_all_info(self):
        print("password:",self.get_password())
        print("balance:",self.get_balance())
        
        


obj=OnlinePayment(1234,1000)
obj.display_all_info()
obj.set_password(12345)
obj.make_payment(200)
obj.send_notification("Payment Successful")
obj.display_all_info()
    
    
    

```

## 22. Define abstract class Readable and abstract class Writable. Create class File thatimplements both.

```python
class Readable(ABC):
    @abstractmethod
    def read(self):
        pass
class Writable(ABC):
    @abstractmethod
    def write(self, data):
        pass
class File(Readable, Writable):
    def read(self):
        print("Reading data from file.")

    def write(self, data):
        print("Writing to file:", data)

f = File()
f.read()
f.write("Hello, Python!")
```

```python
## 27. (Problem) Create abstract class Device with abstract method power_on() and class
Network with method connect(). Create class SmartPhone that inherits from both and
implements the abstract method.
```

```python
from abc import ABC, abstractmethod
class Device(ABC):
    def __init__(self, brand, model):
        self.brand = brand
        self.model = model

    @abstractmethod
    def power_on(self):
        pass

class Network:
    def __init__(self, network_name):
        self.network_name = network_name

    def connect(self):
        print("Connected to network:", self.network_name)


class SmartPhone(Device, Network):
    def __init__(self, brand, model, network_name, price):
        Device.__init__(self, brand, model)
        Network.__init__(self, network_name)
        self.price = price
        
    def power_on(self):
        print(self.brand, self.model, "is powering on...")

    def display_info(self):
        print("SmartPhone Details")
        print("Brand :", self.brand)
        print("Model :", self.model)
        print("Network :", self.network_name)
        print("Price :", self.price)


phone = SmartPhone("Samsung", "Galaxy S25", "5G Network", 80000)

phone.display_info()
phone.power_on()
phone.connect()
```
