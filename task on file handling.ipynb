## 13. Write a program to reverse the content of a file and save it in another file.

```python
file = open("details.txt", "r")
data = file.read()
file.close()
reverse_data = data[::-1]

file = open("data_file.txt", "w")
file.write(reverse_data)
file.close()

print("Reversed content saved successfully.")
```

## 15. Read a file and display the longest word in it.

```python
file=open("data.txt","r")
data=file.read()
words=data.split()
longest=words[0]
for word in words:
    if len(word) > len(longest):
        longest = word
print("longest word:",longest)
file.close
```
