## 1. Plot monthly Revenue as a line chart. Add title, axis labels, markers, grid, and legend.

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
df=pd.read_csv("C:\\Users\\LENOVO\\Downloads\\matplotlib_sales_clean_1000.csv")

df['Order_Date'] = pd.to_datetime(df['Order_Date'])
df['Month'] = df['Order_Date'].dt.strftime('%b')

# Calculate monthly revenue
months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun',
          'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec']

monthly_revenue = df.groupby('Month')['Revenue'].sum()

revenue = [monthly_revenue.get(month, 0) for month in months]

# Plot
plt.plot(months, revenue, marker='o', label='Revenue')

plt.title('Monthly Revenue')
plt.xlabel('Month')
plt.ylabel('Revenue')
plt.grid(True)
plt.legend()

plt.show()
```

## 2. Plot monthly Revenue and Profit on the same chart using different line styles.

```python
df.head(2)
```

```python
df['Order_Date'] = pd.to_datetime(df['Order_Date'])
df['Month'] = df['Order_Date'].dt.month

monthly = df.groupby('Month')[['Revenue', 'Profit']].sum()
plt.plot(monthly.index, monthly['Revenue'], marker='o', label='Revenue')
plt.plot(monthly.index, monthly['Profit'], marker='s', linestyle='--', label='Profit')

plt.title('Monthly Revenue and Profit')
plt.xlabel('Month')
plt.ylabel('Amount')
plt.grid(True)
plt.legend()

plt.show()
```

# 3. Create a bar chart of total Revenue by Product_Category and sort the categoriesfrom highest to lowest revenue.

```python
df = pd.read_csv(r"C:\Users\LENOVO\Downloads\matplotlib_sales_clean_1000.csv")

# Total Revenue for each Product Category
category_revenue = df.groupby('Product_Category')['Revenue'].sum()

# Sort from highest to lowest
category_revenue = category_revenue.sort_values(ascending=False)

# Create bar chart
plt.bar(category_revenue.index, category_revenue.values)

plt.title('Total Revenue by Product Category')
plt.xlabel('Product Category')
plt.ylabel('Total Revenue')
plt.xticks(rotation=45)

plt.show()
```
