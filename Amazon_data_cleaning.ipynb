```python
## clean Ssd_Storage

df['Ssd_Storage'].unique()
```

```python
df['Ssd_Storage'].isnull().sum()
```

```python
def convert_storage(x):
    if pd.isna(x):
        return x
    if "TB" in x:
        return float(x.replace("TB", "")) * 1024
    return float(x.replace("GB", ""))

df["Ssd_Storage"] = df["Ssd_Storage"].apply(convert_storage)
df["Ssd_Storage"]
```

```python
df["Ssd_Storage"] = df["Ssd_Storage"].fillna(df["Ssd_Storage"].mean())
df["Ssd_Storage"]
```

```python
df["Ssd_Storage"] = df["Ssd_Storage"].astype(int)
df["Ssd_Storage"] 
```
